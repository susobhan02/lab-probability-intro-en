What is the chance of rain during next Saturday?

## BONUS

You roll 2 dice. What is the probability that the sum of the numbers is 7? And 8? And 9? Then simulate 10000 times and draw a countplot.

your explanation here

When rolling 2 standard dice, there are a total of **36 possible outcomes** ($6 \times 6$).

*   **Sum is 7:** 
    *   Favorable outcomes: (1,6), (2,5), (3,4), (4,3), (5,2), (6,1) $\rightarrow$ **6 outcomes**
    *   Probability = $$\frac{6}{36} = \mathbf{\frac{1}{6} \approx 16.67\%}$$

*   **Sum is 8:** 
    *   Favorable outcomes: (2,6), (3,5), (4,4), (5,3), (6,2) $\rightarrow$ **5 outcomes**
    *   Probability = $$\mathbf{\frac{5}{36} \approx 13.89\%}$$

*   **Sum is 9:** 
    *   Favorable outcomes: (3,6), (4,5), (5,4), (6,3) $\rightarrow$ **4 outcomes**
    *   Probability = $$\frac{4}{36} = \mathbf{\frac{1}{9} \approx 11.11\%}$$

```python
# YOUR CODE HERE
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# 1. Simulate rolling two dice 10,000 times
np.random.seed(50)  # Ensures reproducible results or any other integer 
die1 = np.random.randint(1, 7, size=10000)
die2 = np.random.randint(1, 7, size=10000)
sums = die1 + die2

# 2. Print empirical probabilities from simulation
print(f"Simulated Probability for 7: {np.mean(sums == 7):.4f} (Expected: 0.1667)")
print(f"Simulated Probability for 8: {np.mean(sums == 8):.4f} (Expected: 0.1389)")
print(f"Simulated Probability for 9: {np.mean(sums == 9):.4f} (Expected: 0.1111)")

# 3. Draw the countplot
plt.figure(figsize=(8, 5))
sns.countplot(x=sums, hue=sums, legend=False, palette="crest")
plt.title("Simulation of Rolling Two Dice 10,000 Times", fontsize=14)
plt.xlabel("Sum of the Numbers", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```
